```python
def get_embedding_index(batch_pert1_n_2:list):
    idx_mat = []
    for p1_p2 in batch_pert1_n_2:
        idxs = [unique_token.index(p) for p in p1_p2]
        idx_mat.append(idxs)
    idx_mat = np.array(idx_mat)
    return torch.from_numpy(idx_mat).long()
```
